```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

path = "./dataset/iris_w_header.csv"
#types = ['f8', 'f8', 'f8', 'f8', 'U50']
data = np.genfromtxt(path, dtype=None, skip_header=1, delimiter=",", usecols = (0,1,2,3), encoding=None)

df = pd.read_csv(path)

fig, ax = plt.subplots(figsize=(5, 5))

ax.scatter(df['sepal_length'], df['petal_length'])

ax.set_xlabel('Petal Length')
ax.set_ylabel('Sepal Length')

print(data.shape)
print("\n", df.describe(), "\n") # statistic information of data usually used for Data Science Application
print(df.head(3))
ax.set_title('Petal Length vs Sepal Length')
plt.show()
```
