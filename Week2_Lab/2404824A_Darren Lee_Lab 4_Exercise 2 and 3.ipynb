```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

## Load csv data and display top 5 data

fishcatch_data = pd.read_csv('dataset/fishcatch.csv')

fishcatch_data.head(5)
```

```python
#Select features X and label Y with proposed columns

X = fishcatch_data[['Length1', 'Length2', 'Length3', 'Height', 'Width']]

Y = fishcatch_data['Weight']

#Display the dimension of X and Y

print(X.shape)
print(Y.shape)
```

```python
# Create model and train

lin_reg1 = LinearRegression()
lin_reg1.fit(X, Y)

# Display intercept and coefficients

print('Intercept: \n', lin_reg1.intercept_)
print('Coefficients: \n', lin_reg1.coef_)
```

```python
## Make prediction

Y_pred = lin_reg1.predict(X)

## Calculate R2 value and display

r2_value = r2_score(Y, Y_pred)
print('R-squared Coefficient of determination: %.2f' %r2_value)
```

```python
## Display the predicted data
fishcatch_data['Y_pred'] = Y_pred

fishcatch_data['Sex'] = np.nan

print(fishcatch_data[['Observation', 'Species', 'Weight', 'Length1', 'Length2', 'Length3', 'Height', 'Width', 'Sex', 'Y_pred']].head())
```

```python
#Create two dimensional array called new_data with proposed data

new_data = [[24,28,34,41,15]]

#predict fish weight and display

fish_weight = lin_reg1.predict(new_data)

print('Predicted weight of the fish is %.2f' %fish_weight)
```
